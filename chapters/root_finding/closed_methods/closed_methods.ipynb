## Closed (Bracketting) methods

Closed iterative root finding methods (aka *Bracketting* methods) work when you can *bound* (bracket) the root, and therefore search through reducing the bounds until you've found it. Such methods are called *closed* because you will generally not wander off far from the root.  

Bracketing methods exploits the fact that functions change sign across the roots of a 1-D function (a simple application of the intermediate value theorem). This does not work for some cases (eg: $1/x$) but are sometimes the only ones that will for others (Non-differentiable / pathological functions).:
